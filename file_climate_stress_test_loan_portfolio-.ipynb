# Climate Stress Test of an Agricultural Loan Portfolio
### How much would a severe drought cost a rural lender, and when does it run out of capital?

**Course link:** Climate Vulnerability and Risk Assessment, *Module 4* (risk quantification models), *Module 10* (scenario-based modeling and uncertainty analysis) and *Module 15* (damage and loss estimation, risk financing and adaptation economics). This notebook builds directly on the farm cash-flow sensitivity notebook: the same climate-to-cash-flow logic now scores **every borrower** in a portfolio.

## 1. The case

A Kenyan **SACCO / microfinance lender** has about **1,500 farmer loans** spread across **four agro-ecological zones**: highland, mid-altitude, semi-arid and an irrigated scheme. Repayments depend on harvests, and harvests depend on rainfall. The board asks:

> *If a severe drought hits, how many loans go bad, how much capital do we lose, and how severe does a drought have to be before we breach our minimum capital ratio?*

## 2. Why this is a stress test, not another sensitivity analysis

| | Sensitivity analysis (previous notebook) | Stress test (this notebook) |
|---|---|---|
| Question | Which inputs matter most? | Does the institution survive a severe shock? |
| Unit | One farm | A whole portfolio, with correlated losses |
| Shocks | Small, graded, often one input at a time | Large, joint, in the tail, defined by return periods |
| Result | Ranking of drivers | Losses, capital, breaking points, management options |

The portfolio view matters because **droughts hit many borrowers at the same time**. A lender cannot diversify away a regional drought the way it can diversify away one farmer's bad luck.

## 3. Objectives

1. Build an illustrative **loan portfolio** with realistic variation between borrowers and zones.
2. Design **severe-but-plausible climate scenarios** from rainfall return periods, not arbitrary numbers.
3. Trace the full chain: **climate shock → yield loss → borrower cash flow → default → loss → lender capital**.
4. Follow the portfolio over **three years**, as liquidity buffers run down and arrears build up.
5. Add a **probabilistic view** (Monte Carlo) so the scenarios can be placed in a loss distribution.
6. Test **management actions** (restructuring, seed, index insurance, concentration limit, stop-loss cover).
7. Run a **reverse stress test**: how severe must a drought be to breach minimum capital?
8. Test **how sensitive the stress-test result is** to our own assumptions.

## 5. Glossary

- **DSCR** (debt service coverage ratio): cash available for repayment ÷ repayment due. Below 1.0 means the borrower cannot cover the instalment.
- **PD** (probability of default): chance a borrower defaults in a year.
- **EAD** (exposure at default): outstanding loan balance when default happens.
- **LGD** (loss given default): share of EAD the lender does not recover.
- **Expected loss** = PD × LGD × EAD.
- **NPL ratio**: share of the loan book that is non-performing (defaulted and not yet written off). Here, loans are written off 12 months after default.
- **Capital ratio**: lender's capital ÷ total assets. Regulators set a minimum.
- **Return period**: a "1-in-20" drought has a 5% chance of occurring in any given year.
- **Reverse stress test**: instead of asking "what happens in this scenario?", ask "what scenario makes us fail?"

## 6. Important note on the data

**All data are illustrative.** The zones, borrowers, loan terms, balance sheet, default curve and thresholds are invented to look plausible and to teach the method. They are **not** real SACCO figures and the zone parameters are not official agro-ecological statistics. The minimum capital ratio is a placeholder "in the spirit of" SACCO prudential rules; check your regulator's actual requirements before using it. Replace the inputs with real data (rainfall records, the lender's loan book, its balance sheet) before drawing real conclusions. Step 12 explains how.
